```python
import pandas as pd
```

```python
data = {"path":("asd","asdd","Asdf","Aefdsa"),"label":(1,2,3,4)}
df = pd.DataFrame(data)
df
```

```python
df["path"].values
```

```python
data_set = tf.data.Dataset.from_tensor_slices(
    (df["path"].values, df["label"].values)
)
for data in data_set:
    print(data)
```

```python
# Changing Transformations
data_set = tf.data.Dataset.from_tensor_slices(tf.range(10))
data_set = data_set.repeat(3).batch(7)
for data in data_set:
    print(data)
```

```python
# Create a dataset with all batches multiplied by 2
data_set = tf.data.Dataset.from_tensor_slices(tf.range(10))
data_set = data_set.repeat(3).batch(7)
data_set = data_set.map(lambda x: x*2, num_parallel_calls=tf.data.AUTOTUNE) # x is batch of data
for data in data_set:
    print(data)
```

```python
data_set = tf.data.Dataset.range(10).repeat(2)
data_set = data_set.shuffle(buffer_size=2, seed=42).batch(7)
for data in data_set:
    print(data)
```
