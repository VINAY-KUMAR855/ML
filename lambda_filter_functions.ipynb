### A function with no name

```python
def add(a,b):
    return a+b
```

```python
add(4,5)
```

```python
addition = lambda a,b:a+b
```

```python
addition(3,4)
```

```python
addition(34,65)
```

```python
def even(a):
    if a%2==0:
        return True
```

```python
even(4)
```

```python
even_check = lambda a:a%2==0
```

```python
even_check(6)
```

```python
even_check(7)
```

### filter function

```python
def even(num):
    if num%2==0:
        return True
```

```python
list_num = [2,3,4,5,6,7,8,9]
```

```python
list(filter(even,list_num))
```

```python
list(filter(lambda num:num%2==0, list_num))
```

```python
list(map(lambda num:num%2==0, list_num))
```
