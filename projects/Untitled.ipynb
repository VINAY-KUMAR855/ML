```python
#issue 2
# change cell 22 and 29
df['month'] = df['date'].dt.to_period('M')  # extracts just Year-Month
monthly_df = df.groupby('month')['total_enrollment'].sum().reset_index()
monthly_df['monthly_growth'] = monthly_df['total_enrollment'].pct_change() * 100
monthly_df
```
