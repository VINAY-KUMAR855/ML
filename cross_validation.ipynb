```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/home/majji-vinaykumar/Documents/diabetes.csv")
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
X = df.iloc[:,:-1].values
Y = df.iloc[:,-1].values
```

```python
print(X)
print(Y)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2,
                                                    stratify = Y, random_state=3)
```

```python
models = [LogisticRegression(max_iter=1000),SVC(kernel="linear"),
          KNeighborsClassifier(),RandomForestClassifier()]
```

```python
def compare_models():
    for model in models:
        model.fit(X_train,Y_train)
        Y_pred = model.predict(X_test)
        print(model,"= ",round(accuracy_score(Y_test,Y_pred),2))
    
```

```python
compare_models()
```

```python
model = LogisticRegression(max_iter=1000)
cross_val = cross_val_score(model,X,Y,cv=5)
print(cross_val)
```

```python
def cross_val():
    for model in models:
        cross = cross_val_score(model,X,Y,cv = 5)
        avg_cross = sum(cross)/len(cross)
        avg_cross = avg_cross*100
        print(model,"=",cross)
        print(model,"=",round(avg_cross,2))
        print("------------------")
```
